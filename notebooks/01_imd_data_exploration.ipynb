```python
import xarray as xr

ds = xr.open_dataset("S:/ai-digital-twin/data/raw/imd_rainfall/rainfall2025.nc")

print(ds)
```

```python
print(ds["RAINFALL"])
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 8))

ds["RAINFALL"].isel(TIME=0).plot(
    cmap="Blues",
    vmin=0,
    vmax=50,
    cbar_kwargs={"label": "Rainfall (mm)"}
)

plt.title("IMD Daily Rainfall\n1 January 2007", fontsize=16)
plt.xlabel("Longitude (°E)")
plt.ylabel("Latitude (°N)")

plt.grid(alpha=0.3)

plt.show()
```

```python
import numpy as np


def rainfall_histogram(self):

    import matplotlib.pyplot as plt

    rainfall = self.ds["RAINFALL"].values.flatten()

    rainfall = rainfall[~np.isnan(rainfall)]

    plt.figure(figsize=(10,6))

    plt.hist(rainfall, bins=50)

    plt.title("Rainfall Distribution")

    plt.xlabel("Rainfall (mm)")

    plt.ylabel("Frequency")

    plt.show()
```
